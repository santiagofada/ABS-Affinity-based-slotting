```python
import numpy as np

Q = np.array([
    [0.13,0.45],
    [0.11,0.69],
    [0.13,0.45],
    [0.12,0.61]])

K = np.array([
    [0.14,0.46],
    [0.10,0.72],
    [0.14,0.46],
    [0.12,0.64]])

V = np.array([
    [0.05,0.6],
    [0.02,0.96],
    [0.05,0.60],
    [0.03,0.84]])
```

```python
def softmax(x):
    z = x
    numerator = np.exp(z)
    denominator = np.sum(numerator, axis=-1, keepdims=True)
    return numerator / denominator
```

```python
mult_1 = softmax(Q @ K.T)
print(mult_1)
```

```python
mult_1[0].sum()
```
